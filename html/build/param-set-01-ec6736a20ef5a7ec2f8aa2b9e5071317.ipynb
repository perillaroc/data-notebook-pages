### 位势高度

```python
field = load_field_from_file(
    geps_grib2_orig_file_path,
    parameter="HGT",
    level_type="pl",
    level=100,
    lazy=True
)
assert field.attrs["GRIB_count"] == 1
```

```python
field = load_field_from_file(
    geps_grib2_orig_file_path,
    parameter={"discipline": 0, "parameterCategory": 3, "parameterNumber": 5},
    level_type="pl",
    level=1000,
    lazy=True
)
assert field.attrs["GRIB_count"] == 17
```

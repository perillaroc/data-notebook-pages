# MESO要素清单：31-48

```python
from reki import from_source
```

```python
from data_notebook.data import get_data_file

path = get_data_file("cma-meso-1km")
source = from_source({"kind": "file", "path": str(path)})
path
```

```python
def check_message(ordinal, field):
    location = field.metadata["source"]
    assert location["ordinal"] == ordinal

    with path.open("rb") as original:
        original.seek(location["offset"])
        assert field.read_grib_message() == original.read(location["length"])


def check_messages(ordinals, fields):
    for ordinal, field in zip(ordinals, fields, strict=True):
        check_message(ordinal, field)
```

## 等高面u风

| 要素名 | cemc | wgrib2 | eccodes | GRIB Key | 层次类型 | 层次 | 编号 |
|---|---|---|---|---|---|---|---|
| 等高面u风 | u | UGRD | u | (0, 2, 2) | heightAboveGround | 30, 50, 70 | 31-33 |
| 等高面u风 | u | UGRD | u | (0, 2, 2) | heightAboveGround | 100 | 34 |
| 等高面u风 | u | UGRD | u | (0, 2, 2) | heightAboveGround | 120, 140, 160, 180 | 35-38 |
| 等高面u风 | u | UGRD | u | (0, 2, 2) | heightAboveGround | 200 | 39 |

```python
# 按 CedarKit 要素名
fields = source.sel(
    parameter="u",
    level_type="height_above_ground",
    level=[30, 50, 70, 100, 120, 140, 160, 180, 200],
    level_unit="m",
).all()
check_messages([31, 32, 33, 34, 35, 36, 37, 38, 39], fields)
```

```python
# 按 ecCodes 短名
by_short_name = source.sel(
    native={"grib": {"shortName": "u"}},
    level_type="height_above_ground",
    level=[30, 50, 70, 100, 120, 140, 160, 180, 200],
    level_unit="m",
).all()
check_messages([31, 32, 33, 34, 35, 36, 37, 38, 39], by_short_name)
```

```python
# 按 GRIB2 原生键
by_grib = source.sel(native={"grib": {
    "discipline": 0,
    "parameterCategory": 2,
    "parameterNumber": 2,
    "typeOfFirstFixedSurface": 103,
    "scaleFactorOfFirstFixedSurface": 0,
    "scaledValueOfFirstFixedSurface": [30, 50, 70, 100, 120, 140, 160, 180, 200],
}}).all()
check_messages([31, 32, 33, 34, 35, 36, 37, 38, 39], by_grib)
```

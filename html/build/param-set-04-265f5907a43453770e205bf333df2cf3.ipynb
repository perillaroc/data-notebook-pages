# MESO要素清单：31-48

```python
import numpy as np
import pandas as pd
import xarray as xr

from reki.format.grib.eccodes import load_field_from_file
```

```python
from data_notebook.data import get_data_file

meso_grib2_orig_file_path = get_data_file("cma-meso-1km")
meso_grib2_orig_file_path
```

## 等高面u风

| 要素名 | cemc | wgrib2 | eccodes | GRIB Key | 层次类型 | 层次 | 编号 |
|---|---|---|---|---|---|---|---|
| 等高面u风 | u | UGRD | u | (0, 2, 2) | heightAboveGround | 30, 50, 70 | 31-33 |
| 等高面u风 | u | UGRD | 100u | (0, 2, 2) | heightAboveGround | 100 | 34 |
| 等高面u风 | u | UGRD | u | (0, 2, 2) | heightAboveGround | 120, 140, 160, 180 | 35-38 |
| 等高面u风 | u | UGRD | 200u | (0, 2, 2) | heightAboveGround | 200 | 39 |

```python
field = load_field_from_file(
    meso_grib2_orig_file_path,
    parameter="UGRD",
    level_type="heightAboveGround",
    level=30,
    lazy=True
)
assert field.attrs["GRIB_count"] == 31
```

```python
field = load_field_from_file(
    meso_grib2_orig_file_path,
    parameter="u",
    level_type="heightAboveGround",
    level=50,
    lazy=True
)
assert field.attrs["GRIB_count"] == 32
```

```python
field = load_field_from_file(
    meso_grib2_orig_file_path,
    parameter="100u",
    lazy=True
)
assert field.attrs["GRIB_count"] == 34
```

```python
field = load_field_from_file(
    meso_grib2_orig_file_path,
    parameter={
        "discipline": 0,
        "parameterCategory": 2,
        "parameterNumber": 2,
    },
    level_type="heightAboveGround",
    level=120,
    lazy=True
)
assert field.attrs["GRIB_count"] == 35
```

```python
field = load_field_from_file(
    meso_grib2_orig_file_path,
    parameter="200u",
    lazy=True
)
assert field.attrs["GRIB_count"] == 39
```

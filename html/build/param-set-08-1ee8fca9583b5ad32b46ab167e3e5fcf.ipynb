# GFS要素清单：472-773

```python
from reki import from_source
```

```python
from data_notebook.data import get_data_file

path = get_data_file("cma-gfs")
source = from_source({"kind": "file", "path": str(path)})
path
```

```python
def check_message(ordinal, field):
    location = field.metadata["source"]
    assert location["ordinal"] == ordinal

    with path.open("rb") as original:
        original.seek(location["offset"])
        assert field.read_grib_message() == original.read(location["length"])


def check_messages(ordinals, fields):
    for ordinal, field in zip(ordinals, fields, strict=True):
        check_message(ordinal, field)
```

## 相对湿度

| 要素名 | cemc | wgrib2 | eccodes | GRIB Key | 层次类型 | 层次 | 编号 |
|---|---|---|---|---|---|---|---|
| 相对湿度 | rh | RH | r | (0, 1, 1) | pl | 30层 | 472-501 |

```python
# 按 CedarKit 要素名
fields = source.sel(
    parameter="rh",
    level_type="pressure",
    level=[1000, 850, 70, 10],
    level_unit="hPa",
).all()
check_messages([472, 477, 497, 501], fields)
```

```python
# 按 ecCodes 短名
by_short_name = source.sel(
    native={"grib": {"shortName": "r"}},
    level_type="pressure",
    level=[1000, 850, 70, 10],
    level_unit="hPa",
).all()
check_messages([472, 477, 497, 501], by_short_name)
```

```python
# 按 GRIB2 原生键
by_grib = source.sel(native={"grib": {
    "discipline": 0,
    "parameterCategory": 1,
    "parameterNumber": 1,
    "typeOfFirstFixedSurface": 100,
    "scaleFactorOfFirstFixedSurface": 0,
    "scaledValueOfFirstFixedSurface": [100000, 85000, 7000, 1000],
}}).all()
check_messages([472, 477, 497, 501], by_grib)
```

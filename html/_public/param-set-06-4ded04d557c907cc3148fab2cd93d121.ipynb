# MESO要素清单：154-279

````python
from reki import from_source
````

````python
from data_notebook.data import get_data_file

path = get_data_file("cma-meso-1km")
source = from_source({"kind": "file", "path": str(path)})
path
````

````python
def check_message(ordinal, field):
    location = field.metadata["source"]
    assert location["ordinal"] == ordinal

    with path.open("rb") as original:
        original.seek(location["offset"])
        assert field.read_grib_message() == original.read(location["length"])
````

## 比湿

| 要素名 | cemc | wgrib2 | eccodes | GRIB Key | 层次类型 | 层次 | 编号 |
|---|---|---|---|---|---|---|---|
| 比湿 | q | SPFH | q | (0, 1, 0) | pl | 21层 | 154-174 |

````python
# 按 CedarKit 要素名
field = source.sel(parameter="q", level_type="pressure", level=1000).one()
check_message(154, field)
````

````python
# 按 ecCodes 短名
by_short_name = source.sel(
    native={"grib": {"shortName": "q"}},
    level_type="pressure",
    level=850,
).one()
check_message(159, by_short_name)
````

````python
# 按 GRIB2 原生键
by_grib = source.sel(native={"grib": {
    "discipline": 0,
    "parameterCategory": 1,
    "parameterNumber": 0,
    "typeOfFirstFixedSurface": 100,
    "scaleFactorOfFirstFixedSurface": 0,
    "scaledValueOfFirstFixedSurface": 10000,
}}).one()
check_message(174, by_grib)
````

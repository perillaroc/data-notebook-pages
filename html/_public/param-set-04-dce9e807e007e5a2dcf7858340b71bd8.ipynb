# GFS要素清单：46-63

```python
from reki import from_source
```

```python
from data_notebook.data import get_data_file

path = get_data_file("cma-gfs")
source = from_source({"kind": "file", "path": str(path)})
path
```

```python
def check_message(ordinal, field):
    location = field.metadata["source"]
    assert location["ordinal"] == ordinal

    with path.open("rb") as original:
        original.seek(location["offset"])
        assert field.read_grib_message() == original.read(location["length"])
```

## UGRD/VGRD

### 等高面u风

| 要素名 | cemc | wgrib2 | eccodes | GRIB Key | 层次类型 | 层次 | 编号 |
|---|---|---|---|---|---|---|---|
| 等高面u风 | u | UGRD | u | (0, 2, 2) | heightAboveGround | 30, 50, 70 | 46-48 |
| 等高面u风 | u | UGRD | u | (0, 2, 2) | heightAboveGround | 100 | 49 |
| 等高面u风 | u | UGRD | u | (0, 2, 2) | heightAboveGround | 120, 140, 160, 180 | 50-53 |
| 等高面u风 | u | UGRD | u | (0, 2, 2) | heightAboveGround | 200 | 54 |

```python
# 按 CedarKit 要素名
field = source.sel(parameter="u", level_type="height_above_ground", level=30).one()
check_message(46, field)
```

```python
# 按 ecCodes 短名
by_short_name = source.sel(
    native={"grib": {"shortName": "u"}},
    level_type="height_above_ground",
    level=120,
).one()
check_message(50, by_short_name)
```

```python
# 按 GRIB2 原生键
by_grib = source.sel(native={"grib": {
    "discipline": 0,
    "parameterCategory": 2,
    "parameterNumber": 2,
    "typeOfFirstFixedSurface": 103,
    "scaleFactorOfFirstFixedSurface": 0,
    "scaledValueOfFirstFixedSurface": 200,
}}).one()
check_message(54, by_grib)
```

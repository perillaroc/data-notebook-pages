# GFS要素清单：64-223

```python
from reki import from_source
```

```python
from data_notebook.data import get_data_file

path = get_data_file("cma-gfs")
source = from_source({"kind": "file", "path": str(path)})
path
```

```python
def check_message(ordinal, field):
    location = field.metadata["source"]
    assert location["ordinal"] == ordinal

    with path.open("rb") as original:
        original.seek(location["offset"])
        assert field.read_grib_message() == original.read(location["length"])
```

## HGT/TMP/UGRD/VGRD

### 位势高度

| 要素名 | cemc | wgrib2 | eccodes | GRIB Key | 层次类型 | 层次 | 编号 |
|---|---|---|---|---|---|---|---|
| 位势高度 | h | HGT | gh | (0, 3, 5) | pl | 40层 | 64-103 |


```python
# 按 CedarKit 要素名
field = source.sel(parameter="h", level_type="pressure", level=1000).one()
check_message(64, field)
```

```python
# 用 parameter 按 wgrib2 短名
field = source.sel(parameter="HGT", level_type="pressure", level=850).one()
check_message(69, field)
```

```python
# 用 parameter 按 ecCodes 短名
by_short_name = source.sel(parameter="gh", level_type="pressure", level=1.5).one()
check_message(99, by_short_name)
```

```python
# 按 GRIB2 原生键
by_grib = source.sel(native={"grib": {
    "discipline": 0,
    "parameterCategory": 3,
    "parameterNumber": 5,
    "typeOfFirstFixedSurface": 100,
    "scaleFactorOfFirstFixedSurface": 0,
    "scaledValueOfFirstFixedSurface": 10,
}}).one()
check_message(103, by_grib)
```
